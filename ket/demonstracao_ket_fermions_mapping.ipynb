# ⚛️ Demonstração Prática no Ket: Férmions e Mapeamento para Pauli Strings

Este notebook demonstra o fluxo completo da simulação fermiônica usando o **Ket**:
1. **Criação e Aniquilação individuais** ($a_2^\dagger$ e $a_2$) e sua compilação em Pauli Strings com Jordan-Wigner.
2. **Movimento de elétrons (Hopping $a_0^\dagger a_2$)** com a geração automática da cadeia de paridade $Z_1$.

```python
import sys, os
sys.path.insert(0, os.path.abspath("src"))

import ket
from ket import Process, CreateFermion, AnnihilateFermion, FermionSentence, jordan_wigner, bravyi_kitaev

# 1. Aloca o registrador quântico com 4 qubits
p = Process()
q = p.alloc(4)
```

## 1. Operadores Fundamentais: Criação ($a_2^\dagger$) e Aniquilação ($a_2$)

Veja como o Ket traduz cada operador na cadeia de sensores $Z_0 Z_1$ seguida da rotação local $(X_2 \mp i Y_2)/2$:

```python
a2_dag = FermionSentence({CreateFermion(2): 1.0})
a2_ann = FermionSentence({AnnihilateFermion(2): 1.0})

print("=== 1. Criação no Orbital 2 (a†_2) ===")
print(jordan_wigner(a2_dag, q))

print("\n=== 2. Aniquilação no Orbital 2 (a_2) ===")
print(jordan_wigner(a2_ann, q))
```

## 2. Movimento do Elétron: Salto entre Orbitais (*Hopping* $a_0^\dagger a_2$)

Ao compilar o salto do elétron, o Ket gera as Pauli Strings com portas $X$ e $Y$ nas pontas (0 e 2) e a porta $Z_1$ no meio:

```python
# Termo Hermitiano de salto molecular: ½ (a†_0 a_2 + a†_2 a_0)
hopping_term = FermionSentence({
    CreateFermion(0) * AnnihilateFermion(2): 0.5,
    CreateFermion(2) * AnnihilateFermion(0): 0.5
})

print("=== 3. Operador Fermiônico de Salto ===")
print("Férmions:", hopping_term)

print("\n=== 4. Pauli Strings Geradas (Jordan-Wigner) ===")
H_jw = jordan_wigner(hopping_term, q)
print(H_jw)
```
