# Demonstração de Coeficientes de Orbitais Moleculares (`mo_coeff`)

Este notebook demonstra o comportamento e a origem da matriz de coeficientes `mo_coeff` gerada pelo resolvedor clássico Hartree-Fock (**PySCF**) e lida pelo **Ket**.

### 1. Importações e Configuração da Molécula ($H_2$)

Primeiro, importamos a biblioteca do PySCF e definimos nossa molécula de Hidrogênio ($H_2$).

```python
import pyscf
import numpy as np

# Definição da molécula H2 com base sto-3g
mol = pyscf.gto.M(atom='H 0 0 0; H 0 0 0.74', basis='sto-3g')
print("Molécula construída com sucesso!")
```

### 2. Criação do Resolvedor Hartree-Fock (RHF)

Agora instanciamos o resolvedor clássico RHF. Note que, antes de rodar o cálculo, a matriz de coeficientes `mo_coeff` é `None` (ainda não foi calculada).

```python
mf = pyscf.scf.RHF(mol)

print("O resolvedor possui a variável 'mo_coeff'?", hasattr(mf, 'mo_coeff'))
print("Valor atual do 'mo_coeff':", getattr(mf, 'mo_coeff', None))
```

### 3. Execução da Simulação Clássica

Chamamos o método `.kernel()` para fazer o cálculo iterativo clássico e convergir a função de onda eletrônica.

```python
energia_scf = mf.kernel()
print(f"Energia SCF Convergida: {energia_scf} Hartree")
```

### 4. Análise da Matriz `mo_coeff` pós-convergência

Agora que o cálculo convergiu, a matriz `mo_coeff` foi preenchida pelo PySCF. Vamos imprimi-la e entender as dimensões.

```python
print("Tipo da matriz:", type(mf.mo_coeff))
print("Dimensão da matriz:", mf.mo_coeff.shape)
print("\nConteúdo da matriz de coeficientes moleculares:")
print(mf.mo_coeff)
```

### 5. Explicação Química da Matriz

Como a molécula de $H_2$ com a base `sto-3g` possui 2 orbitais atômicos e gera 2 orbitais moleculares, a matriz de coeficientes é de dimensão **2x2**:

*   **Linha 0:** Contribuição do orbital atômico $1s$ do primeiro átomo de Hidrogênio.
*   **Linha 1:** Contribuição do orbital atômico $1s$ do segundo átomo de Hidrogênio.
*   **Coluna 0 (Primeiro Orbital Molecular - HOMO/Ligante):** Ambas as contribuições são positivas e iguais ($0.5488 \cdot \text{Atom}_A + 0.5488 \cdot \text{Atom}_B$). É o orbital ligante de menor energia onde os elétrons residem.
*   **Coluna 1 (Segundo Orbital Molecular - LUMO/Antiligante):** As contribuições têm sinais opostos ($-1.2124 \cdot \text{Atom}_A + 1.2124 \cdot \text{Atom}_B$). É o orbital antiligante desocupado.
