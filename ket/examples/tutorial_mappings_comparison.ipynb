# Resultados e Comparação Prática: Mapeamentos Fermiônicos para Qubits no Ket

Este notebook compara a saída dos mapeamentos fermiônicos da plataforma **Ket** (`ket.mapping`) com a biblioteca **OpenFermion** (`openfermion`):
1. **Jordan-Wigner (JW)** — Ket vs. OpenFermion
2. **Bravyi-Kitaev Padrão (BK)** — Ket vs. OpenFermion
3. **Bravyi-Kitaev com Simetria (SCBK)** — Hopping a⁺_0 a_2 (Redução de 4 -> 2 qubits)
4. **BK com Simetria (SCBK)** — Operador Número Total N (Redução de 4 -> 2 qubits)
5. **Paridade (Parity)** — Implementação nativa do Ket

## 1. Mapeamento Jordan-Wigner (JW) — Ket vs. OpenFermion

Comparação no termo de acoplamento distante a⁺_0 a_2 em N=4 spin-orbitais.

```python
import sys, os
sys.path.insert(0, os.path.abspath("../src"))
import ket
from ket import Process, CreateFermion, AnnihilateFermion, FermionSentence
from ket.mapping import jordan_wigner, parity, bravyi_kitaev, symmetry_conserving_bravyi_kitaev
import openfermion

p = Process()
q = p.alloc(4)

ket_hopping = FermionSentence({CreateFermion(0) * AnnihilateFermion(2): 1.0})
of_hopping = openfermion.FermionOperator("0^ 2", 1.0)

print("=== 1. JORDAN-WIGNER (JW) ===")
print("Ket JW:")
print(jordan_wigner(ket_hopping, q))
print()
print("OpenFermion JW:")
print(openfermion.jordan_wigner(of_hopping))
```

## 2. Mapeamento Bravyi-Kitaev Padrão (BK) — Ket vs. OpenFermion

Comparação no termo a⁺_0 a_2 em N=4 spin-orbitais.

```python
print("=== 2. BRAVYI-KITAEV (BK) ===")
print("Ket BK:")
print(bravyi_kitaev(ket_hopping, q))
print()
print("OpenFermion BK:")
print(openfermion.bravyi_kitaev(of_hopping))
```

## 3. Bravyi-Kitaev com Simetria (SCBK) no Hopping — Ket vs. OpenFermion

Redução de 4 para 2 qubits no termo a⁺_0 a_2 (N=4 orbitais, 2 elétrons).

```python
ket_scbk_hopping = symmetry_conserving_bravyi_kitaev(ket_hopping, q, active_orbitals=4, n_alpha=1, n_beta=1)
of_scbk_hopping = openfermion.symmetry_conserving_bravyi_kitaev(of_hopping, active_orbitals=4, active_fermions=2)

print("=== 3. BRAVYI-KITAEV COM SIMETRIA (SCBK) - Hopping a⁺_0 a_2 ===")
print("Ket SCBK (Reduzido para 2 qubits):")
print(ket_scbk_hopping)
print()
print("OpenFermion SCBK (Reduzido para 2 qubits):")
print(of_scbk_hopping)
```

## 4. Operador Número Total N em BK com Simetria (SCBK) — Ket vs. OpenFermion

Comparação do operador número total N em 4 orbitais com redução para 2 qubits.

```python
ket_n_tot = ket.number_operator(4)
of_n_tot = sum(openfermion.FermionOperator(f"{i}^ {i}") for i in range(4))

ket_scbk_n = symmetry_conserving_bravyi_kitaev(ket_n_tot, q, active_orbitals=4, n_alpha=1, n_beta=1)
of_scbk_n = openfermion.symmetry_conserving_bravyi_kitaev(of_n_tot, active_orbitals=4, active_fermions=2)

print("=== 4. OPERADOR NÚMERO TOTAL N EM BK COM SIMETRIA (SCBK) ===")
print("Ket SCBK (Reduzido para 2 qubits):")
print(ket_scbk_n)
print()
print("OpenFermion SCBK (Reduzido para 2 qubits):")
print(of_scbk_n)
```

## 5. Mapeamento por Paridade (Parity) — Ket

Mapeamento por Paridade nativo do Ket para o termo a⁺_0 a_2 e para o Operador Número Total N.

```python
print("=== 5. PARIDADE (PARITY) NO KET ===")
print("Ket Parity - Hopping a⁺_0 a_2 (4 qubits):")
print(parity(ket_hopping, q))
print()
print("Ket Parity - Operador Número Total N (4 qubits):")
print(parity(ket_n_tot, q))
```
